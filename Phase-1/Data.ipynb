### <b>Sequence Operations</b>
* Access index from 0 to length of list - 1
* Slicing takes the sublist from it

```python
# indexing
fruits = ['apple','banana','oranges']
print(fruits[0],fruits[len(fruits)-1])
print(fruits[-len(fruits)] , fruits[-1])
```

```python
# Slicing 
fruits = ["apples","oranges","grapes","guava","lana"]
print(fruits[:3])
print(fruits[2:])
print(fruits[1:4])
```

```python
# Concatenation and Repetition
numList = [1,2,23]
print(numList *2)
print(numList + [3,4])
```

## <b> Dictionary</b>
* Not Sequence
* Mapping labels like key-value pairs

```python
person = {
    "name": "karen",
    "age" : 32,
    "gender" :"male"
}
print(person)
print(person['name'])
```

```python
# key-value pairs
fruits = {}
fruits['apple'] = 180
fruits['orange'] = 60
print(fruits)
```

```python
# Creating Dictionary
students = dict(Raven=32,May=24)
print(students)
person = dict(zip(['name','age'],['maya',24]))
print(person)
```

```python
# Nesting
person = {
    'name':{
        'first':'San',
        'last':'Altman'
    }
}
person['name']['first']
```

```python
# append the element in dictionary
person = {}
person['name'] = 'karen'
# key is not present default value
print(person.get("salary",24500))
```

```python
person = {}
print(person['name'])# Key Error
```

```python
person = {'age':32}
if 'name' not in person:
    print("Key is not found")
```

* keys : generator of keys
* values : generator of values
* items : generator of key-value pairs

```python
person = {
    "name": "karen",
    "age" : 32,
    "gender" :"male"
}
print(person.keys())
print(person.values())
print(person.items())
```

### <b> Iteration using for loops </b>

```python
for key in person.keys():
    print(key)
for key in person:
    print(key)
for val in person.values():
    print(val)
for key,val in person.items():
    print(key,val)
```

## <b>Sets</b>

```python
# empty set
empty_set = set()
print(empty_set)

# set
# add and remove
A ={1,2,3}
A.add(4)
A.add(3)
print(A)
A.remove(4)
print(A)
A.remove(12)
```

```python
# set operations
# union
A ={1,2,3}
B ={2,3,4}
print(A | B)
# intersection
A = {1,2,3,4}
B = {2,3,5,6}
print(A & B)
```

```python
# difference
A ={1,2,3,5,6}
B = {2,3,4}
print(A - B)
# superset
A ={1,2,3,4,5,6,7,8}
B ={2,3,4,5,6}
print(A > B) # super set
```
